## Synthetic Data

```python
import numpy as np

np.random.seed(42)

X = np.random.randn(100, 5)

weights = np.random.randn(X.shape[1])

proba = X @ weights.T + 0.5 + np.random.randn(X.shape[0])

sigmoid = 1 / (1 + (np.exp(-proba)))

y = np.where(sigmoid >= 0.5, 1, 0)

y
```

```python
import pandas as pd

X = pd.DataFrame(X)

X.head()
```

```python
model = Gradient_Boosting_Classifier(max_depth=2, n_estimators=100)

model.fit(X, y)
```

```python
model.estimators[:5]
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X)

accuracy_score(y, y_pred)
```

## Titanic

```python
import pandas as pd

X_train_preprocessed = pd.read_csv('../data/titanic/preprocessed/train.csv').set_index('PassengerId')
X_val_preprocessed = pd.read_csv('../data/titanic/preprocessed/val.csv').set_index('PassengerId')
y_train = pd.read_csv('../data/titanic/preprocessed/y_train.csv').set_index('PassengerId').squeeze('columns')
y_val = pd.read_csv('../data/titanic/preprocessed/y_val.csv').set_index('PassengerId').squeeze('columns')

X_train_preprocessed.head()
```

### Model Training

```python
model = Gradient_Boosting_Classifier()
model.fit(X_train_preprocessed, y_train)
```

```python
model.estimators[:5]
```

```python
y_pred = model.predict(X_val_preprocessed)

accuracy_score(y_val, y_pred)
```

#### SciKit-Learn

```python
from sklearn.ensemble import GradientBoostingClassifier

model_sklearn = GradientBoostingClassifier(n_estimators=50, max_depth=5, random_state=42)

model_sklearn.fit(X_train_preprocessed, y_train)
```

```python
y_pred_sklearn = model_sklearn.predict(X_val_preprocessed)

accuracy_score(y_val, y_pred_sklearn)
```

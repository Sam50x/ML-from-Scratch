## House Prices

```python
import pandas as pd

X_train_preprocessed = pd.read_csv('../data/house_prices/preprocessed/train.csv').set_index('Id')
X_val_preprocessed = pd.read_csv('../data/house_prices/preprocessed/val.csv').set_index('Id')
y_train = pd.read_csv('../data/house_prices/preprocessed/y_train.csv').set_index('Id').squeeze('columns')
y_val = pd.read_csv('../data/house_prices/preprocessed/y_val.csv').set_index('Id').squeeze('columns')

X_train_preprocessed.head()
```

### Model Training

#### My Model

```python
model = Gradient_Boosting_Regressor(n_estimators=50)
model.fit(X_train_preprocessed, y_train)
```

```python
model.estimators[:5]
```

```python
from sklearn.metrics import root_mean_squared_error

y_val_pred = model.predict(X_val_preprocessed)

rmse_val = root_mean_squared_error(y_val_pred, y_val)

rmse_val
```

#### SciKit-Learn

```python
from sklearn.ensemble import GradientBoostingRegressor

model_sklearn = GradientBoostingRegressor(max_depth=5, n_estimators=50, learning_rate=0.1, random_state=42)

model_sklearn.fit(X_train_preprocessed, y_train)
```

```python
y_val_pred_sklearn = model_sklearn.predict(X_val_preprocessed)

rmse_val_sklearn = root_mean_squared_error(y_val_pred_sklearn, y_val)

rmse_val_sklearn
```

### Retraining and Submission

```python
X_full_preprocessed = pd.read_csv('../data/house_prices/preprocessed/full.csv').set_index('Id')
y = pd.read_csv('../data/house_prices/preprocessed/y_full.csv').set_index('Id').squeeze('columns')

X_full_preprocessed.head()
```

```python
model.fit(X_full_preprocessed, y)
```

```python
model_sklearn.fit(X_full_preprocessed, y)
```

```python
X_test_preprocessed = pd.read_csv('../data/house_prices/preprocessed/test.csv').set_index('Id')

X_test_preprocessed.head()
```

```python
y_test = model.predict(X_test_preprocessed)
y_test_sklearn = model_sklearn.predict(X_test_preprocessed)
```

```python
submission_df = pd.DataFrame(y_test, columns=['SalePrice'], index=X_test_preprocessed.index)

submission_df.to_csv('../data/house_prices/submissions/submission_boosting_my_model.csv', sep=',')
```

```python
submission_sklearn_df = pd.DataFrame(y_test_sklearn, columns=['SalePrice'], index=X_test_preprocessed.index)

submission_sklearn_df.to_csv('../data/house_prices/submissions/submission_boosting_sklearn.csv', sep=',')
```

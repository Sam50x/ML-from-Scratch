## XM

```python
import pandas as pd

df = pd.read_csv('../data/xm/data.csv').set_index('ID')

df.head()
```

```python
df.info()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df_scaled = scaler.fit_transform(df)

df_scaled = pd.DataFrame(df_scaled, index=df.index, columns=df.columns)

df_scaled.head()
```

### Model Training

#### My Model

```python
from sklearn.metrics import silhouette_score

model = GMM(n_clusters=4)

labels = model.fit_predict(df_scaled)

silhouette_score(df_scaled, labels)
```

#### Scikit-Learn Model

```python
from sklearn.mixture import GaussianMixture

model_sklearn = GaussianMixture(
    n_components=4,
    reg_covar=1e-6,
    n_init=1,
    init_params='random',
    random_state=42
)

labels_sklearn = model_sklearn.fit_predict(df_scaled)

silhouette_score(df_scaled, labels_sklearn)
```

### Submission

```python
submission = pd.DataFrame(labels, index=df.index, columns=['Category'])

submission.to_csv('../data/xm/submissions/submission_gmm_my_model.csv')
```

```python
submission_sklearn = pd.DataFrame(labels_sklearn, index=df.index, columns=['Category'])

submission_sklearn.to_csv('../data/xm/submissions/submission_gmm_sklearn.csv')
```
